```
# === Cell 1: common setup ===
import os, glob, json, warnings, zipfile
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from joblib import Parallel, delayed
warnings.filterwarnings("ignore")
plt.rcParams.update({"font.size": 11, "font.family": "serif",
                     "axes.grid": True, "grid.alpha": 0.3})
N_JOBS = os.cpu_count() or 2
OUT = "/kaggle/working" if os.path.isdir("/kaggle/working") else "./out"
os.makedirs(OUT, exist_ok=True)
T = 72
print(f"CPU cores: {N_JOBS} | output directory: {OUT}")
```

```
# === Cell 2: core module — generator (v2), detectors, classifiers ===

def _ar1_noise(n, rng, rho=0.5, sigma=0.15):
    e = np.zeros(n)
    for t in range(1, n):
        e[t] = rho * e[t-1] + rng.normal(0, sigma)
    return e

def generate_series(typology, rng, censor=True, burst_prob=0.5):
    """typology in {'null','endogenous','exogenous'}; log10 scale.
    v2: a fraction of endogenous series is burst-initiated (Simge archetype)."""
    base = rng.uniform(1.0, 3.0)
    onset = int(rng.integers(18, T-24))
    lead_yt, lead_tr = int(rng.integers(1,4)), int(rng.integers(1,3))
    L = np.full(T, base, float)
    if typology == "endogenous":
        amp = rng.uniform(1.0, 2.0); k = rng.uniform(0.6, 1.4)
        ramp = amp/(1+np.exp(-k*(np.arange(T)-onset-6)))
        ds = onset + int(rng.integers(8,14)); d = rng.uniform(0.06,0.12)
        plateau = rng.uniform(0.35,0.6); decay = np.ones(T)
        for t in range(ds, T):
            decay[t] = plateau + (1-plateau)*np.exp(-d*(t-ds))
        L = base + ramp*decay
    elif typology == "exogenous":
        amp = rng.uniform(1.5,2.5); d = rng.uniform(0.25,0.5)
        for t in range(onset, T):
            L[t] = base + amp*np.exp(-d*(t-onset))
    sy = lead_yt if typology=="endogenous" else 0
    st = lead_tr if typology=="endogenous" else 0
    L_yt = np.concatenate([L[sy:], np.full(sy, L[-1])])
    L_tr = np.concatenate([L[st:], np.full(st, L[-1])])
    yt = L_yt + _ar1_noise(T, rng)
    tr0 = L_tr + _ar1_noise(T, rng)
    spu = L + _ar1_noise(T, rng)
    if typology=="endogenous" and rng.random() < burst_prob:
        # burst-initiated endogenous: 30-70% of eventual amplitude within 2 months
        for arr in (yt, tr0, spu):
            b = arr[:onset].mean(); amp2 = arr[onset:].max()-b
            frac = rng.uniform(0.3, 0.7)
            for t in range(onset, min(onset+2, T)):
                arr[t] = max(arr[t], b + frac*amp2 + rng.normal(0, 0.05))
    if censor:
        c = base + rng.uniform(0.5, 0.9)
        sp = np.where(spu >= c, spu, 0.0)
    else:
        sp = spu
    lin = 10**tr0
    tr = np.log10(np.maximum(1, np.round(100*lin/lin.max())))
    return dict(typology=typology, onset=onset, sp=sp, yt=yt, tr=tr,
                sp_uncensored=spu)

# ---------- causal flags ----------
def causal_flags(x, thr, burn_in=12, censored=False, persist=2):
    x = np.asarray(x, float); raw = np.zeros(len(x), bool)
    for t in range(burn_in, len(x)):
        past = x[:t]
        if censored:
            past = past[past > 0]
            if len(past) < 6 or x[t] == 0: continue
        sd = past.std() or 1e-6
        if x[t] > past.mean() + thr*sd: raw[t] = True
    out = np.zeros_like(raw)
    for t in range(persist-1, len(raw)):
        if raw[max(0,t-persist+1):t+1].sum() >= persist: out[t] = True
    return out

def cusum_flags(x, k=0.5, h=6.0, burn_in=12, censored=False):
    x = np.asarray(x, float); Cst = 0.0; out = np.zeros(len(x), bool)
    for t in range(burn_in, len(x)):
        past = x[:t]
        if censored:
            past = past[past > 0]
            if len(past) < 6 or x[t] == 0: continue
        sd = past.std() or 1e-6
        Cst = max(0.0, Cst + (x[t]-past.mean())/sd - k)
        if Cst > h: out[t] = True
    return out

def composite_flags(s, thr, burn_in=12, persist=2):
    """Causal CP-PSI: mean of the three expanding-window z-scores (single series)."""
    n = len(s["yt"]); comp = np.zeros(n)
    for t in range(burn_in, n):
        zs = []
        for x, cens in [(s["yt"],False),(s["tr"],False),(s["sp"],True)]:
            past = x[:t]
            if cens:
                past = past[past>0]
                if len(past)<6 or x[t]==0: continue
            sd = past.std() or 1e-6
            zs.append((x[t]-past.mean())/sd)
        comp[t] = np.mean(zs) if zs else 0.0
    raw = comp > thr
    out = np.zeros_like(raw)
    for t in range(persist-1, n):
        if raw[max(0,t-persist+1):t+1].sum() >= persist: out[t] = True
    return out

def first_confirmed(allf, confirm, window=2):
    for t in range(allf.shape[1]):
        lo = max(0, t-window+1)
        if allf[:, lo:t+1].any(axis=1).sum() >= confirm: return t
    return None

def detect_pe(s, thr=1.75, confirm=2, window=2, persist=2):
    allf = np.vstack([causal_flags(s["yt"],thr,persist=persist),
                      causal_flags(s["tr"],thr,persist=persist),
                      causal_flags(s["sp"],thr,censored=True,persist=persist)])
    return first_confirmed(allf, confirm, window)

def detect_cusum(s, k=0.5, h=6.0, confirm=2, window=2):
    allf = np.vstack([cusum_flags(s["yt"],k,h), cusum_flags(s["tr"],k,h),
                      cusum_flags(s["sp"],k,h,censored=True)])
    return first_confirmed(allf, confirm, window)

def detect_composite(s, thr=1.75, persist=2):
    f = composite_flags(s, thr, persist=persist)
    return int(f.argmax()) if f.any() else None

# ---------- regime classification ----------
def rise_ratio(yt, ign, h=12):
    b = yt[max(0,ign-6):ign].mean()
    peak = yt[ign:min(ign+h+1,len(yt))].max()
    return (yt[ign]-b)/max(peak-b, 1e-6)

def causal_features(s, ign, h):
    yt, tr, sp = s["yt"], s["tr"], s["sp"]
    b = yt[max(0,ign-6):ign].mean()
    w = yt[ign:min(ign+h+1,len(yt))]
    def jz(x, c=False):
        p = x[:ign]; p = p[p>0] if c else p
        return abs(x[ign]-p[-1])/(p.std() or 1e-6) if len(p)>5 else 0.0
    trb = tr[max(0,ign-6):ign].mean()
    tramp = max(tr[ign:min(ign+h+1,len(tr))].max()-trb, 1e-6)
    return [rise_ratio(yt,ign,h), (tr[ign]-trb)/tramp,
            (w.max()-yt[ign])/max(w.max()-b,1e-6),
            np.argmax(w)/max(h,1),
            (w[-1]-w.max())/max(w.max()-b,1e-6),
            jz(yt), jz(tr), jz(sp,True),
            yt[max(0,ign-12):ign].std()]
print("core module loaded")
```

## EXP1 — Operating-characteristic grid (persistent-exceedance detector)

```
# === Cell 3: EXP1 ===
def run_cell(thr, confirm, persist, censor=True, n_per_class=300, seed=7,
             detector="pe", cusum_k=0.5, cusum_h=6.0):
    rng = np.random.default_rng(seed)
    rows = []
    for typ in ["null","endogenous","exogenous"]:
        for _ in range(n_per_class):
            s = generate_series(typ, rng, censor=censor)
            if detector == "pe":
                ign = detect_pe(s, thr, confirm, persist=persist)
            elif detector == "cusum":
                ign = detect_cusum(s, k=cusum_k, h=cusum_h, confirm=confirm)
            else:
                ign = detect_composite(s, thr, persist=persist)
            valid = ign is not None and typ != "null" and ign >= s["onset"]-3
            regime = None
            if valid:
                regime = "exogenous" if rise_ratio(s["yt"], ign, 12) >= 0.6 \
                         else "endogenous"
            rows.append(dict(truth=typ, det=ign is not None, valid=valid,
                             prem=(ign is not None and typ!="null"
                                   and ign < s["onset"]-3),
                             regime=regime,
                             delay=(ign-s["onset"]) if valid else np.nan,
                             lead=(int(np.argmax(s["sp"]))-ign) if valid else np.nan))
    df = pd.DataFrame(rows)
    ev, nu = df[df.truth!="null"], df[df.truth=="null"]
    ok = ev[ev.valid]
    return dict(detector=detector, thr=thr, confirm=confirm, persist=persist,
                censor=censor, TPR=100*ev.valid.mean(),
                premature=100*ev.prem.mean(), FPR=100*nu.det.mean(),
                delay=ok.delay.mean(), lead=ok.lead.mean(),
                regime_acc=100*(ok.regime==ok.truth).mean())

grid_cfg = [(thr,c,p) for thr in [1.25,1.5,1.75,2.0,2.25,2.5]
            for c in [1,2,3] for p in [1,2]]
grid = Parallel(n_jobs=N_JOBS)(delayed(run_cell)(thr,c,p) for thr,c,p in grid_cfg)
g = pd.DataFrame(grid).round(2)
g.to_csv(f"{OUT}/results_grid.csv", index=False)
print(g.to_string(index=False))
```

## EXP4 — Regime classification: horizon sweep and sim-to-real transfer
Rise ratio (single parameter) vs. logistic regression vs. gradient boosting.
All three are trained/evaluated on the synthetic testbed; their zero-shot
transfer to the two real archetypes is assessed in EXP5. Both outcomes of the
sim-to-real comparison are reported as-is in the article.

```
# === Cell 7: EXP4 ===
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

def gen_labeled(n, seed):
    rng = np.random.default_rng(seed); out = []
    i = 0
    while len(out) < n:
        typ = "endogenous" if i%2==0 else "exogenous"; i += 1
        s = generate_series(typ, rng)
        ign = detect_pe(s, 1.75, 2, persist=2)
        if ign is None or ign < s["onset"]-3: continue
        out.append((s, ign, typ))
    return out

train = gen_labeled(1600, seed=21)
test  = gen_labeled(800,  seed=99)
rows4 = []
for h in [3, 6, 9, 12]:
    Xtr = [causal_features(s,i,h) for s,i,_ in train]
    ytr = [t for *_, t in train]
    Xte = [causal_features(s,i,h) for s,i,_ in test]
    yte = np.array([t for *_, t in test])
    rr = np.array(["exogenous" if f[0]>=0.6 else "endogenous" for f in Xte])
    lr = make_pipeline(StandardScaler(),
                       LogisticRegression(max_iter=1000)).fit(Xtr, ytr)
    gb = GradientBoostingClassifier(random_state=0).fit(Xtr, ytr)
    rows4.append(dict(horizon=h,
                      rise=100*(rr==yte).mean(),
                      logreg=100*(lr.predict(Xte)==yte).mean(),
                      gbm=100*(gb.predict(Xte)==yte).mean()))
    if h == 12: LR12, GB12 = lr, gb   # keep for the real-case cells
r4 = pd.DataFrame(rows4).round(1)
r4.to_csv(f"{OUT}/results_regime_horizon.csv", index=False)
print(r4.to_string(index=False))
fig, ax = plt.subplots(figsize=(6,3.8))
for col, lbl in [("rise","rise-ratio"),("logreg","LogReg"),("gbm","GBM")]:
    ax.plot(r4.horizon, r4[col], marker="o", label=lbl)
ax.set_xlabel("confirmation horizon h (months)")
ax.set_ylabel("regime accuracy (%)"); ax.legend()
ax.set_title("Regime identifiability vs confirmation horizon (synthetic)")
fig.tight_layout(); fig.savefig(f"{OUT}/fig_regime_horizon.png", dpi=200)
plt.show()
```

## EXP5 — Zero-shot validation on the real archetypes

```
# === Cell 8: EXP5 ===
def find(fname):
    hits = glob.glob(f"/kaggle/input/**/{fname}", recursive=True) or \
           glob.glob(f"./**/{fname}", recursive=True)
    assert hits, f"{fname} not found - attach the dataset via Add Input"
    return hits[0]

def load_real(fp, spc, ytc, trc):
    df = pd.read_csv(fp, parse_dates=["final_date"]).set_index("final_date")
    spv = df[spc].fillna(0).to_numpy(float)
    return df.index, dict(sp=np.where(spv>0, np.log10(spv+1), 0.0),
                          yt=np.log10(df[ytc].to_numpy(float)),
                          tr=np.log10(df[trc].to_numpy(float)))

cases = [("Simge - Askin Olayim", "FINAL_ACADEMIC_DATASET_SIMGE.csv",
          ("tr_streams","yt_views","google_search_interest"), "endogenous"),
         ("Kate Bush - RUTH", "FINAL_ACADEMIC_DATASET_KATE_BUSH.csv",
          ("sp_streams","yt_views","trends_interest"), "exogenous")]
rows5, plots = [], []
for name, fname, cols, truth in cases:
    idx, s = load_real(find(fname), *cols)
    ign = detect_pe(s, 1.75, 2, persist=2)
    rr = rise_ratio(s["yt"], ign, 12)
    regime = "exogenous" if rr >= 0.6 else "endogenous"
    f12 = causal_features(s, ign, 12)
    lead = int(np.argmax(s["sp"])) - ign
    rows5.append(dict(case=name, truth=truth,
                      ignition=str(idx[ign].date()),
                      regime_rise=regime, rise_ratio=round(rr,3),
                      regime_logreg=LR12.predict([f12])[0],
                      regime_gbm=GB12.predict([f12])[0],
                      lead_to_peak_mo=lead))
    plots.append((idx, s, ign, name, regime, rr))
r5 = pd.DataFrame(rows5)
r5.to_csv(f"{OUT}/results_real_cases.csv", index=False)
print(r5.to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))
for ax, (idx, s, ign, name, regime, rr) in zip(axes, plots):
    ax.plot(idx, s["yt"], label="YouTube (log10)", lw=1.4)
    ax.plot(idx, s["tr"], label="Trends (log10)", lw=1.4)
    ax.plot(idx, np.where(s["sp"]>0, s["sp"], np.nan),
            label="Spotify (censored)", lw=1.4)
    ax.axvline(idx[ign], color="red", ls="--", lw=1.5,
               label=f"Ignition {idx[ign].strftime('%Y-%m')}")
    ax.set_title(f"{name}\nclassified: {regime} (r={rr:.2f})", fontsize=10)
    ax.legend(fontsize=8, loc="upper left")
fig.tight_layout(); fig.savefig(f"{OUT}/fig_real_cases.png", dpi=200)
plt.show()
```

## EXP6 — Honest forecasting benchmark (expanding-window, one-step-ahead)

```
# === Cell 9: EXP6 ===
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score, mean_absolute_error

def expanding_bench(fname, spc, ytc, trc, label, min_train=24):
    df = pd.read_csv(find(fname), parse_dates=["final_date"]).set_index("final_date")
    d = df["2021-01-01":].copy()
    y_raw = np.log1p(d[spc].fillna(0).astype(float))
    m = pd.DataFrame({"y": y_raw})
    for l in [1,2,3]: m[f"yt{l}"] = np.log1p(d[ytc]).shift(l)
    for l in [1,2]: m[f"tr{l}"] = np.log1p(d[trc]).shift(l)
    m["sp1"] = m["y"].shift(1)
    m = m.dropna()
    X, y = m.drop(columns="y"), m["y"]
    preds = {"SVR": [], "RF": [], "Ridge": [], "Naive": []}; idxs = []
    for t in range(min_train, len(m)):
        Xtr, ytr, xt = X.iloc[:t], y.iloc[:t], X.iloc[[t]]
        preds["SVR"].append(SVR(kernel="rbf", C=8.0, epsilon=0.1)
                            .fit(Xtr, ytr).predict(xt)[0])
        preds["RF"].append(RandomForestRegressor(150, max_depth=6,
                           random_state=42).fit(Xtr, ytr).predict(xt)[0])
        preds["Ridge"].append(Ridge().fit(Xtr, ytr).predict(xt)[0])
        preds["Naive"].append(y.iloc[t-1]); idxs.append(t)
    yt_ = y.iloc[idxs]
    return [dict(case=label, model=k, R2=round(r2_score(yt_, v),3),
                 MAE=round(mean_absolute_error(yt_, v),3))
            for k, v in preds.items()]

bench = (expanding_bench("FINAL_ACADEMIC_DATASET_SIMGE.csv","tr_streams",
                         "yt_views","google_search_interest","Simge (endo)") +
         expanding_bench("FINAL_ACADEMIC_DATASET_KATE_BUSH.csv","sp_streams",
                         "yt_views","trends_interest","Kate Bush (exo)"))
b = pd.DataFrame(bench)
b.to_csv(f"{OUT}/results_forecast_bench.csv", index=False)
print(b.to_string(index=False))
```

## EXP4b + EXP7 — delta-jump baseline and multi-seed statistics
- **EXP4b:** The single-jump delta rule of the original IVPM (2.0 sigma) and a
  fair variant with the cut tuned on the training set, evaluated under the same
  test conditions as rise-ratio / LogReg / GBM.
- **EXP7:** Replication over 5 seeds; mean +/- sd, 95% CI (t distribution, n=5),
  within-run paired **McNemar** tests, and across-seed **Wilcoxon**
  signed-rank tests.

```
# === Cell 11: EXP4b tanimlar — delta-sicrama siniflandiricisi ===
from scipy.stats import binomtest, wilcoxon, t as tdist

SEEDS = [7, 17, 27, 37, 47]
N_TRAIN, N_TEST = 1600, 800

def delta_jump(s, ign):
    """Maximum causal z-jump at ignition (the original IVPM statistic)."""
    best = 0.0
    for x, cens in [(s["yt"],False),(s["tr"],False),(s["sp"],True)]:
        p = x[:ign]; p = p[p>0] if cens else p
        if len(p) > 5:
            best = max(best, abs(x[ign]-p[-1])/(p.std() or 1e-6))
    return best

def ci95(v):
    v = np.asarray(v, float); n = len(v)
    h = tdist.ppf(0.975, n-1)*v.std(ddof=1)/np.sqrt(n)
    return v.mean(), v.std(ddof=1), h

def mcnemar_p(correct_a, correct_b):
    b = int(np.sum(correct_a & ~correct_b))
    c = int(np.sum(~correct_a & correct_b))
    return (binomtest(min(b,c), b+c).pvalue if b+c>0 else 1.0), b, c
print("EXP4b/EXP7 definitions ready | seeds:", SEEDS)
```

```
# === Cell 12: EXP7a — multi-seed operating points ===
cfg7a = ([(th,2,2,True) for th in [1.25,1.5,1.75,2.0,2.25,2.5]] +
         [(1.75,1,2,True),(1.75,3,2,True),(1.75,2,2,False),(1.75,3,2,False)])
jobs = [(th,c,p,cen,sd) for (th,c,p,cen) in cfg7a for sd in SEEDS]
res7a = Parallel(n_jobs=N_JOBS)(
    delayed(run_cell)(th, c, p, censor=cen, seed=sd) for th,c,p,cen,sd in jobs)
sg = pd.DataFrame(res7a)
sg["seed"] = [sd for *_, sd in jobs]
sg.to_csv(f"{OUT}/results_seed_grid.csv", index=False)

def agg(gdf):
    out = {}
    for m in ["TPR","FPR","premature","delay","lead","regime_acc"]:
        mu, sdv, h = ci95(gdf[m].values)
        out[m] = f"{mu:.2f} ± {sdv:.2f} (±{h:.2f})"
    return pd.Series(out)
stats7a = (sg.groupby(["thr","confirm","persist","censor"])
             .apply(agg).reset_index())
stats7a.to_csv(f"{OUT}/results_seed_grid_stats.csv", index=False)
print("Multi-seed statistics [mean ± sd (± 95% CI)]:")
print(stats7a.to_string(index=False))
```

```
# === Cell 13: EXP7b — regime classification: 5 seeds + EXP4b + McNemar ===
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

def regime_seed_run(sd):
    train = gen_labeled(N_TRAIN, seed=sd)
    test  = gen_labeled(N_TEST,  seed=sd+1000)
    yte = np.array([t for *_, t in test])
    out = dict(seed=sd)
    # delta baseline (independent of h)
    dj_tr = np.array([delta_jump(s,i) for s,i,_ in train])
    ytr = np.array([t for *_, t in train])
    dj_te = np.array([delta_jump(s,i) for s,i,_ in test])
    pred_d20 = np.where(dj_te >= 2.0, "exogenous", "endogenous")
    cuts = np.arange(0.5, 4.01, 0.25)
    accs = [( (np.where(dj_tr>=c,"exogenous","endogenous")==ytr).mean(), c)
            for c in cuts]
    best_cut = max(accs)[1]
    pred_dtn = np.where(dj_te >= best_cut, "exogenous", "endogenous")
    out["delta20"] = 100*(pred_d20==yte).mean()
    out["delta_tuned"] = 100*(pred_dtn==yte).mean()
    out["delta_best_cut"] = best_cut
    corr = {"delta20": pred_d20==yte, "delta_tuned": pred_dtn==yte}
    for h in [3, 6, 12]:
        Xtr = [causal_features(s,i,h) for s,i,_ in train]
        Xte = [causal_features(s,i,h) for s,i,_ in test]
        pr = np.array(["exogenous" if f[0]>=0.6 else "endogenous" for f in Xte])
        lr = make_pipeline(StandardScaler(),
                           LogisticRegression(max_iter=1000)).fit(Xtr, ytr)
        gb = GradientBoostingClassifier(random_state=0).fit(Xtr, ytr)
        out[f"rise_h{h}"]   = 100*(pr==yte).mean()
        out[f"logreg_h{h}"] = 100*(lr.predict(Xte)==yte).mean()
        out[f"gbm_h{h}"]    = 100*(gb.predict(Xte)==yte).mean()
        if h == 12:
            corr["rise"] = pr==yte
            corr["gbm"]  = gb.predict(Xte)==yte
    # within-run paired McNemar tests (h=12)
    for a, b in [("rise","delta20"), ("rise","delta_tuned"), ("gbm","rise")]:
        p, nb_, nc_ = mcnemar_p(corr[a], corr[b])
        out[f"mcnemar_{a}_vs_{b}_p"] = p
    return out

res7b = Parallel(n_jobs=min(N_JOBS, len(SEEDS)))(
    delayed(regime_seed_run)(sd) for sd in SEEDS)
sr = pd.DataFrame(res7b)
sr.to_csv(f"{OUT}/results_seed_regime.csv", index=False)
print(sr.round(2).to_string(index=False))

print("\n--- Summary [mean ± sd (± 95% CI)] ---")
for col in ["rise_h3","rise_h6","rise_h12","logreg_h12","gbm_h12",
            "delta20","delta_tuned"]:
    mu, sdv, h = ci95(sr[col].values)
    print(f"  {col:12s}: {mu:5.1f} ± {sdv:4.1f} (±{h:4.1f})")

print("\n--- Across-seed Wilcoxon (n=5; attainable minimum p = 0.0625) ---")
for a, b in [("rise_h12","delta20"), ("rise_h12","delta_tuned"),
             ("gbm_h12","rise_h12"), ("rise_h12","rise_h3")]:
    try:
        wp = wilcoxon(sr[a], sr[b]).pvalue
    except ValueError:
        wp = float("nan")
    print(f"  {a} vs {b}: diff={sr[a].mean()-sr[b].mean():+.1f} points, p={wp:.4f}")
print("\nMcNemar (within-run, h=12) median p-values:")
for c in [c for c in sr.columns if c.startswith("mcnemar")]:
    print(f"  {c}: median p = {sr[c].median():.2e}")
```

## EXP3b — Fully replicated detector comparison (Revision R1.9 / R2.5)

Persistent exceedance, CUSUM, causal CP-PSI, and Bayesian online
change-point detection (BOCPD, Gaussian-mean variant) are all evaluated
inside the same 2-of-3 architecture, with 5 seeds × 900 series each.
Results replace the single-run screen of Table 6 in the original paper.

```
# === EXP3b: fully replicated detector comparison (R1.9 / R2.5) ===
from scipy.stats import norm as _norm_sp

def bocpd_flags(x, lam=36, burn_in=12, warmup=6, censored=False, thresh=0.15):
    """
    BOCPD with Normal-Normal conjugate model (Adams & MacKay 2007).
    Prior calibrated from burn-in; flags when P(run_length=0 | data) > thresh.
    lam=36: expected run length 3 years (suitable for 72-month series).
    thresh=0.15: tuned operating point.
    NOTE: Kalman updates track gradual ramps quickly, so BOCPD is less
    sensitive to endogenous (logistic) events than to exogenous (step) events.
    This is reported as a finding, not treated as a failure to fix.
    """
    x = np.asarray(x, float); n = len(x); raw = np.zeros(n, bool)
    burn = x[:burn_in]
    if censored: burn = burn[burn > 0]
    mu0    = float(burn.mean()) if len(burn) >= 3 else 0.0
    sigma0 = max(float(burn.std()) if len(burn) >= 3 else 1.0, 0.10)
    s_obs  = sigma0; hazard = 1.0 / lam
    lw       = np.array([0.0])
    mu_post  = np.array([mu0])
    var_post = np.array([sigma0**2])
    active   = 0
    for t in range(burn_in, n):
        xt = x[t]
        if censored and xt == 0: continue
        pred_var   = var_post + s_obs**2
        log_pred   = _norm_sp.logpdf(xt, mu_post, np.sqrt(pred_var))
        log_growth = lw + log_pred + np.log(1.0 - hazard)
        log_cp     = (np.logaddexp.reduce(lw)
                      + _norm_sp.logpdf(xt, mu0,
                                        np.sqrt(sigma0**2 + s_obs**2))
                      + np.log(hazard))
        new_lw = np.append(log_cp, log_growth)
        new_lw -= np.logaddexp.reduce(new_lw)
        # Save P(run_length=0) BEFORE pruning
        p_cp_now = float(np.exp(new_lw[0]))
        K         = var_post / (var_post + s_obs**2)
        mu_post   = np.append(mu0,       mu_post  + K * (xt - mu_post))
        var_post  = np.append(sigma0**2, (1.0 - K) * var_post)
        lw = new_lw
        if len(lw) > 40:
            keep     = np.argsort(lw)[-40:]
            lw       = lw[keep]
            mu_post  = mu_post[keep]
            var_post = var_post[keep]
            lw      -= np.logaddexp.reduce(lw)
        active += 1
        if active > warmup and p_cp_now > thresh:
            raw[t] = True
    return raw


def detect_bocpd(s, confirm=2, window=3, lam=36, thresh=0.15):
    """2-of-3 BOCPD detector. window=3 allows slightly asynchronous channel flags."""
    allf = np.vstack([
        bocpd_flags(s["yt"], lam=lam, thresh=thresh),
        bocpd_flags(s["tr"], lam=lam, thresh=thresh),
        bocpd_flags(s["sp"], lam=lam, censored=True, thresh=thresh),
    ])
    return first_confirmed(allf, confirm, window)


def _run_detector_seed(det_fn, seed, n_per_class):
    rng = np.random.default_rng(seed)
    tp = fp = fn = delays = 0
    for typ in ("endogenous", "exogenous", "null"):
        for _ in range(n_per_class):
            s   = generate_series(typ, rng)
            ign = det_fn(s)
            true_on = s["onset"]
            if typ == "null":
                if ign is not None: fp += 1
            else:
                # valid: not more than 3 months early; no upper bound
                if ign is not None and ign >= true_on - 3:
                    tp += 1; delays += max(0, ign - true_on)
                else:
                    fn += 1
    tpr   = tp / (2 * n_per_class) * 100
    fa    = fp / n_per_class * 100
    delay = delays / max(tp, 1)
    return tpr, fa, delay


SEEDS_3b = [7, 17, 27, 37, 47]
N_3b     = 300

detectors = {
    "PE (theta=1.75)":    lambda s: detect_pe(s, 1.75, 2, persist=2),
    "CUSUM (h=8)":        lambda s: detect_cusum(s, h=8, confirm=2),
    "Composite (t=2.0)":  lambda s: detect_composite(s, thr=2.0, persist=2),
    "BOCPD (lam=36)":     lambda s: detect_bocpd(s),
}

rows_3b = []
for det_name, det_fn in detectors.items():
    print(f"  Running {det_name}...", end=" ", flush=True)
    results = Parallel(n_jobs=N_JOBS)(
        delayed(_run_detector_seed)(det_fn, seed, N_3b)
        for seed in SEEDS_3b
    )
    tprs, fas, dls = zip(*results)
    rows_3b.append(dict(
        detector   = det_name,
        tpr_mean   = np.mean(tprs),  tpr_std   = np.std(tprs, ddof=1),
        fa_mean    = np.mean(fas),    fa_std    = np.std(fas,  ddof=1),
        delay_mean = np.mean(dls),    delay_std = np.std(dls,  ddof=1),
    ))
    print(f"TPR={np.mean(tprs):.1f}±{np.std(tprs,ddof=1):.1f}"
          f"  FA={np.mean(fas):.1f}±{np.std(fas,ddof=1):.1f}")

# BOCPD regime breakdown (exo vs endo)
print("\n  BOCPD regime breakdown:")
for typ in ("endogenous", "exogenous"):
    rng_b = np.random.default_rng(7)
    tp_b = fn_b = 0
    for _ in range(500):
        s   = generate_series(typ, rng_b)
        ign = detect_bocpd(s)
        if ign is not None and ign >= s["onset"] - 3: tp_b += 1
        else: fn_b += 1
    print(f"    {typ:12s}: TPR={tp_b/500*100:.1f}%")

df_3b = pd.DataFrame(rows_3b)
df_3b.to_csv(f"{OUT}/results_detector_replicated.csv", index=False)

# Figure: TPR-FA plane with error bars
fig, ax = plt.subplots(figsize=(7, 5))
colors  = ["#2b6cb0", "#e07b39", "#38a169", "#805ad5"]
markers = ["o", "s", "^", "D"]
for i, row in df_3b.iterrows():
    ax.errorbar(row.fa_mean, row.tpr_mean,
                xerr=row.fa_std, yerr=row.tpr_std,
                fmt=markers[i % 4], color=colors[i % 4],
                capsize=5, ms=8, lw=2, label=row.detector)
ax.set_xlabel("False-alarm rate (%)")
ax.set_ylabel("Valid TPR (%)")
ax.set_title("Fully replicated detector comparison\n"
             "(5 seeds \u00d7 900 series, all inside 2-of-3 architecture)")
ax.legend(fontsize=9)
fig.tight_layout()
fig.savefig(f"{OUT}/fig_detector_replicated.png", dpi=180, bbox_inches="tight")
plt.close()
print("fig_detector_replicated.png saved")
```

## EXP_sensitivity — Generator parameter sensitivity analysis (Revision R2.3)

Each generator parameter is varied one at a time while holding all others at
their default values. We report TPR and FA at the selected operating point
(theta=1.75, m=2, p=2) for each parameter level, using 300 series per class
per configuration (single seed=7 for speed; main results use 5 seeds).

```
# === EXP_sensitivity: generator parameter sensitivity (R2.3) ===

def _tpr_fa_config(gen_kwargs, n=300, seed=7, thr=1.75, confirm=2, persist=2):
    """Run one configuration: return (TPR%, FA%)."""
    rng = np.random.default_rng(seed)
    tp = fp = fn = 0
    for typ in ('endogenous','exogenous','null'):
        for _ in range(n):
            s = generate_series(typ, rng, **gen_kwargs)
            ign = detect_pe(s, thr, confirm, persist=persist)
            true_on = s['onset']
            if typ == 'null':
                if ign is not None: fp += 1
            else:
                if ign is not None and ign >= true_on - 3:  # valid: not more than 3 mo. early
                    tp += 1
                else:
                    fn += 1
    tpr = tp / (2*n) * 100
    fa  = fp / n * 100
    return tpr, fa

# Parameter sweep grid
# Format: (label, kwarg_overrides)
# We monkey-patch generate_series defaults via wrapper for each sweep.
import copy

def _gen_with_override(typ, rng, **ov):
    """Call generate_series but override specific rng draws via fixed seeds trick.
    Simpler approach: vary global T or pass a modified rng factory.
    Here we use a thin wrapper that adjusts the draw ranges via closure overrides.
    """
    # Override specific parameters by substituting deterministic values
    # into the generate_series internals is complex; instead, we vary
    # the parameters that can be passed directly.
    return generate_series(typ, rng, **ov)

sweeps = []

# 1. Endogenous amplitude range
for amp_lo, amp_hi, lbl in [(0.5,1.5,'amp endo [0.5,1.5]'),
                             (1.0,2.0,'amp endo [1.0,2.0] (default)'),
                             (1.5,2.5,'amp endo [1.5,2.5]')]:
    # Cannot directly override amp without modifying generate_series.
    # We record the label and note it requires generate_series modification.
    sweeps.append((lbl, None, 'amp_endo'))

# Because generate_series draws internally, we implement a local patched version:
def _gen_patched(typ, rng, censor=True,
                 amp_endo_lo=1.0, amp_endo_hi=2.0,
                 amp_exo_lo=1.5,  amp_exo_hi=2.5,
                 censor_lo=0.5,   censor_hi=0.9,
                 burst_prob=0.5,  lead_max_yt=3,
                 ar_rho=0.5,      ar_sigma=0.15):
    base = rng.uniform(1.0, 3.0)
    onset = int(rng.integers(18, T-24))
    lead_yt = int(rng.integers(1, lead_max_yt+1))
    lead_tr = int(rng.integers(1, 3))
    L = np.full(T, base, float)
    if typ == 'endogenous':
        amp = rng.uniform(amp_endo_lo, amp_endo_hi)
        k = rng.uniform(0.6, 1.4)
        ramp = amp/(1+np.exp(-k*(np.arange(T)-onset-6)))
        ds = onset + int(rng.integers(8, 14)); d = rng.uniform(0.06, 0.12)
        plateau = rng.uniform(0.35, 0.6); decay = np.ones(T)
        for t in range(ds, T):
            decay[t] = plateau + (1-plateau)*np.exp(-d*(t-ds))
        L = base + ramp*decay
    elif typ == 'exogenous':
        amp = rng.uniform(amp_exo_lo, amp_exo_hi)
        d = rng.uniform(0.25, 0.5)
        for t in range(onset, T):
            L[t] = base + amp*np.exp(-d*(t-onset))
    sy = lead_yt if typ == 'endogenous' else 0
    st = lead_tr if typ == 'endogenous' else 0
    L_yt = np.concatenate([L[sy:], np.full(sy, L[-1])])
    L_tr = np.concatenate([L[st:], np.full(st, L[-1])])
    def _ar1(n):
        e = np.zeros(n)
        for t in range(1, n):
            e[t] = ar_rho*e[t-1] + rng.normal(0, ar_sigma)
        return e
    yt = L_yt + _ar1(T); tr0 = L_tr + _ar1(T); spu = L + _ar1(T)
    if typ == 'endogenous' and rng.random() < burst_prob:
        for arr in (yt, tr0, spu):
            b = arr[:onset].mean(); amp2 = arr[onset:].max()-b
            frac = rng.uniform(0.3, 0.7)
            for t in range(onset, min(onset+2, T)):
                arr[t] = max(arr[t], b + frac*amp2 + rng.normal(0, 0.05))
    if censor:
        c = base + rng.uniform(censor_lo, censor_hi)
        sp = np.where(spu >= c, spu, 0.0)
    else:
        sp = spu
    lin = 10**tr0
    tr = np.log10(np.maximum(1, np.round(100*lin/lin.max())))
    return dict(typology=typ, onset=onset, sp=sp, yt=yt, tr=tr, sp_uncensored=spu)

def _tpr_fa_patched(n=300, seed=7, thr=1.75, confirm=2, persist=2, **gen_kw):
    rng = np.random.default_rng(seed)
    tp = fp = 0
    for typ in ('endogenous','exogenous','null'):
        for _ in range(n):
            s = _gen_patched(typ, rng, **gen_kw)
            ign = detect_pe(s, thr, confirm, persist=persist)
            if typ == 'null':
                if ign is not None: fp += 1
            else:
                true_on = s['onset']
                if ign is not None and ign >= true_on - 3:  # valid: not more than 3 mo. early
                    tp += 1
    return tp/(2*n)*100, fp/n*100

SENS_GRID = [
    # label, kwarg_overrides
    # Endogenous amplitude
    ('amp_endo [0.5,1.5]',  dict(amp_endo_lo=0.5,  amp_endo_hi=1.5)),
    ('amp_endo [1.0,2.0]',  dict()),  # default
    ('amp_endo [1.5,2.5]',  dict(amp_endo_lo=1.5,  amp_endo_hi=2.5)),
    # Exogenous amplitude
    ('amp_exo [1.0,2.0]',   dict(amp_exo_lo=1.0,   amp_exo_hi=2.0)),
    ('amp_exo [1.5,2.5]',   dict()),  # default
    ('amp_exo [2.0,3.0]',   dict(amp_exo_lo=2.0,   amp_exo_hi=3.0)),
    # Censoring threshold
    ('censor_lo=0.3,hi=0.7', dict(censor_lo=0.3,   censor_hi=0.7)),
    ('censor_lo=0.5,hi=0.9', dict()),  # default
    ('censor_lo=0.7,hi=1.1', dict(censor_lo=0.7,   censor_hi=1.1)),
    # AR noise
    ('AR sigma=0.08',        dict(ar_sigma=0.08)),
    ('AR sigma=0.15',        dict()),  # default
    ('AR sigma=0.25',        dict(ar_sigma=0.25)),
    # Platform lead (endo)
    ('lead_yt_max=1',        dict(lead_max_yt=1)),
    ('lead_yt_max=3',        dict()),  # default
    ('lead_yt_max=5',        dict(lead_max_yt=5)),
    # Burst probability
    ('burst_prob=0.0',       dict(burst_prob=0.0)),
    ('burst_prob=0.5',       dict()),  # default
    ('burst_prob=1.0',       dict(burst_prob=1.0)),
]

rows_sens = []
print("Running generator sensitivity sweep...")
for lbl, kw in SENS_GRID:
    tpr, fa = _tpr_fa_patched(n=300, seed=7, **kw)
    rows_sens.append(dict(config=lbl, TPR=tpr, FA=fa))
    print(f"  {lbl:30s}  TPR={tpr:.1f}  FA={fa:.1f}")

df_sens = pd.DataFrame(rows_sens)
df_sens.to_csv(f"{OUT}/results_sensitivity.csv", index=False)

# Figure: TPR and FA per configuration
fig, axes = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
axes[0].bar(range(len(df_sens)), df_sens.TPR, color='#2b6cb0')
axes[0].axhline(98.8, color='red', ls='--', label='default TPR=98.8')
axes[0].set_ylabel('TPR (%)'); axes[0].legend(fontsize=9)
axes[0].set_title('Generator sensitivity: TPR and FA at theta=1.75, m=2, p=2')
axes[1].bar(range(len(df_sens)), df_sens.FA, color='#e07b39')
axes[1].axhline(2.1, color='red', ls='--', label='default FA=2.1')
axes[1].set_ylabel('FA (%)'); axes[1].legend(fontsize=9)
axes[1].set_xticks(range(len(df_sens)))
axes[1].set_xticklabels(df_sens.config, rotation=45, ha='right', fontsize=8)
fig.tight_layout()
fig.savefig(f"{OUT}/fig_sensitivity.png", dpi=180, bbox_inches='tight')
plt.close(); print("fig_sensitivity.png saved")
```

## EXP_missingness — Ablation under broader missingness mechanisms (Revision R2.4)

The chart-censoring operator is replaced with three alternative mechanisms to
test whether the 2-of-3 structural-necessity finding generalises:
1. **MCAR** – each streaming observation masked with probability p_miss.
2. **Threshold-shift** – the censoring threshold varies over time (increasing).
3. **Partial censoring** – only a random fraction of below-threshold values
   are set to zero (softer truncation).

```
# === EXP_missingness: ablation under broader missingness (R2.4) ===

def _gen_mcar(typ, rng, p_miss=0.5):
    """Replace chart-censoring with MCAR on the streaming channel."""
    s = generate_series(typ, rng, censor=False)
    sp = s['sp'].copy()
    mask = rng.random(len(sp)) < p_miss
    sp[mask] = 0.0
    s['sp'] = sp
    return s

def _gen_threshold_shift(typ, rng, shift_rate=0.02):
    """Threshold rises linearly over time (increasing censoring)."""
    s = generate_series(typ, rng, censor=False)
    base = s['sp_uncensored'][:12].mean()
    c0 = base + 0.5
    thresholds = c0 + shift_rate * np.arange(T)
    sp = np.where(s['sp_uncensored'] >= thresholds, s['sp_uncensored'], 0.0)
    s['sp'] = sp
    return s

def _gen_partial_censor(typ, rng, censor_prob=0.7):
    """Below-threshold values zeroed with probability censor_prob (soft)."""
    s = generate_series(typ, rng, censor=False)
    spu = s['sp_uncensored']
    base = spu[:12].mean()
    c = base + rng.uniform(0.5, 0.9)
    below = spu < c
    zero_mask = below & (rng.random(T) < censor_prob)
    sp = np.where(zero_mask, 0.0, spu)
    s['sp'] = sp
    return s

def _ablation_run(gen_fn, n=300, seed=7):
    rng = np.random.default_rng(seed)
    res = {}
    for confirm in (2, 3):
        tp = fp = fn = 0
        for typ in ('endogenous','exogenous','null'):
            for _ in range(n):
                s = gen_fn(typ, rng)
                ign = detect_pe(s, 1.75, confirm, persist=2)
                if typ == 'null':
                    if ign is not None: fp += 1
                else:
                    true_on = s['onset']
                    if ign is not None and ign >= true_on - 3:  # valid: not more than 3 mo. early
                        tp += 1
                    else:
                        fn += 1
        res[f'{confirm}of3_TPR'] = tp/(2*n)*100
        res[f'{confirm}of3_FA']  = fp/n*100
    return res

SEEDS_MISS = [7, 17, 27, 37, 47]
mechanisms = [
    ('Chart censoring (default)',   lambda t,r: generate_series(t,r,censor=True)),
    ('MCAR (p_miss=0.50)',          lambda t,r: _gen_mcar(t,r,p_miss=0.50)),
    ('MCAR (p_miss=0.70)',          lambda t,r: _gen_mcar(t,r,p_miss=0.70)),
    ('Threshold-shift (rate=0.02)', lambda t,r: _gen_threshold_shift(t,r,0.02)),
    ('Partial censor (prob=0.70)',  lambda t,r: _gen_partial_censor(t,r,0.70)),
]

rows_miss = []
print("Running missingness ablation...")
for mech_name, gen_fn in mechanisms:
    seed_res = Parallel(n_jobs=N_JOBS)(
        delayed(_ablation_run)(gen_fn, n=300, seed=s) for s in SEEDS_MISS
    )
    for confirm in (2, 3):
        tprs = [r[f'{confirm}of3_TPR'] for r in seed_res]
        fas  = [r[f'{confirm}of3_FA']  for r in seed_res]
        rows_miss.append(dict(
            mechanism = mech_name,
            rule      = f'{confirm}-of-3',
            TPR_mean  = np.mean(tprs),
            TPR_std   = np.std(tprs, ddof=1),
            FA_mean   = np.mean(fas),
            FA_std    = np.std(fas, ddof=1),
        ))
    print(f"  {mech_name}")

df_miss = pd.DataFrame(rows_miss)
df_miss.to_csv(f"{OUT}/results_missingness.csv", index=False)
print(df_miss.to_string(index=False))

# Figure
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
mech_names = df_miss[df_miss.rule=='2-of-3']['mechanism'].tolist()
x = np.arange(len(mech_names)); width = 0.35
tpr2 = df_miss[df_miss.rule=='2-of-3']['TPR_mean'].values
tpr3 = df_miss[df_miss.rule=='3-of-3']['TPR_mean'].values
err2 = df_miss[df_miss.rule=='2-of-3']['TPR_std'].values
err3 = df_miss[df_miss.rule=='3-of-3']['TPR_std'].values
axes[0].bar(x-width/2, tpr2, width, label='2-of-3', color='#2b6cb0',
            yerr=err2, capsize=4)
axes[0].bar(x+width/2, tpr3, width, label='3-of-3', color='#e07b39',
            yerr=err3, capsize=4)
axes[0].set_xticks(x); axes[0].set_xticklabels(mech_names, rotation=30, ha='right')
axes[0].set_ylabel('TPR (%)'); axes[0].legend()
axes[0].set_title('TPR: 2-of-3 vs 3-of-3 under different missingness mechanisms')
fa2 = df_miss[df_miss.rule=='2-of-3']['FA_mean'].values
fa3 = df_miss[df_miss.rule=='3-of-3']['FA_mean'].values
axes[1].bar(x-width/2, fa2, width, color='#2b6cb0', label='2-of-3')
axes[1].bar(x+width/2, fa3, width, color='#e07b39', label='3-of-3')
axes[1].set_xticks(x); axes[1].set_xticklabels(mech_names, rotation=30, ha='right')
axes[1].set_ylabel('FA (%)'); axes[1].legend()
axes[1].set_title('False-alarm rate under different missingness mechanisms')
fig.tight_layout()
fig.savefig(f"{OUT}/fig_missingness.png", dpi=180, bbox_inches='tight')
plt.close(); print("fig_missingness.png saved")

# Revizyon ciktilari zip'e ekle
rev_files = ['results_cut_sensitivity.csv', 'results_detector_replicated.csv',
             'results_sensitivity.csv', 'results_missingness.csv',
             'fig_cut_sensitivity.png', 'fig_detector_replicated.png',
             'fig_sensitivity.png', 'fig_missingness.png']
zp = f"{OUT}/ivpm_ew_results_v3.zip"
with zipfile.ZipFile(zp, 'w', zipfile.ZIP_DEFLATED) as zf:
    for fn in os.listdir(OUT):
        if fn.endswith(('.csv', '.png')):
            zf.write(os.path.join(OUT, fn), fn)
print("Bundled:", zp)
```
